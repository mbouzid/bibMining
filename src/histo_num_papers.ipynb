# Description

This notebook generates a histogram of the number of papers published per year in the corpus. The output file is in EPS format.

# Requirements

An environment with:
- matplotlib
- pandas
- openpyxl


Use the following to install the packages:

```pip install -e ".[histo_num_papers]"```

# Prerequisites

- An excel file ```filename``` containing the references in ```sheet_name``` with at least the column ```column_name```.

````python
import pandas as pd
from matplotlib.ticker import MaxNLocator
import matplotlib.pyplot as plt
````

````python
filename = '../data/refs.xlsx'
sheet_name = 'RAW'
output_filename = 'histo_num_papers.eps'
column_name = 'Year'
````

````python
df = pd.read_excel(filename, sheet_name=sheet_name)
````

````python
year_counts = df[column_name].value_counts().sort_index()
````

````python
plt.figure(figsize=(10,4))
plt.bar(year_counts.index, year_counts.values, color='skyblue', edgecolor='black')

plt.ylabel('Number of papers')
plt.gca().xaxis.set_major_locator(MaxNLocator(integer=True))
plt.gca().yaxis.set_major_locator(MaxNLocator(integer=True))

plt.xticks(year_counts.index,fontsize=12)
plt.savefig(output_filename, format="eps", bbox_inches="tight")
plt.show()
````
