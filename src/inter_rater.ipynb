# Description

This code computes the inter-rater reliability statistics.

# Prerequisites

- metrics


Use the following to install the packages:

```pip install -e ".[inter_rater]"```


# Input

You need a `filename` in excel format in the `data` folder.
The `sheets` ('Indicators', 'ResilienceBarriers','ResilienceEnablers') must contain at least the following columns:
- 'Id'
- 'rater1'
- 'rater2'

# Output 

- Derived cohen's kappas of each sheet

````python
import pandas as pd
from sklearn.metrics import cohen_kappa_score
````

````python
filename = "../data/Rating-Dimensions.xlsx"
````

````python
sheets=["Indicators","ResilienceBarriers","ResilienceEnablers"]
````

````python
for sheet in sheets:
    df = pd.read_excel(filename, sheet_name=sheet)  # columns: rater1, rater2
    kappa = cohen_kappa_score(df["rater1"], df["rater2"])
    print(f"{sheet} Cohen's κ for {sheet} = {kappa:.3f}")
````

````python
df = pd.read_excel(filename,sheet_name=sheet)  # columns: rater1, rater2
kappa = cohen_kappa_score(df["rater1"], df["rater2"])
print(f"Cohen's κ = {kappa:.3f}")
````
